## 1) Create Metadata: Write metadata files for the following datasets:

```python
import pandas as pd

baseline_data = pd.read_csv("data/baseline_data.csv")
blood_data = pd.read_csv("data/blood_data.csv")
diag_data = pd.read_csv("data/diag_data.csv")
dict_data = pd.read_csv("data/dict_data.csv")
events = pd.read_csv("data/events.csv")
```

```python
baseline_data
```

```python
print("-"*100)
print("Metadata DataFrame for: baseline_data")
print("-"*100)
baseline_metadata_df = pd.DataFrame({
    'Data Type': baseline_data.dtypes,
}).reset_index().rename(columns={'index': 'Column Name'})

baseline_descriptions = {
    "id": "Unique identifier for each patient",
    "sex": "Biological sex of the patient",
    "d.birth": "Patient's date of birth",
    "smoking": "Smoking status of the patient (current, former, never)"
}
baseline_metadata_df['Description'] = baseline_metadata_df['Column Name'].map(baseline_descriptions)
print(baseline_metadata_df.to_string())


# Metadata for dict_data
print("-"*100)
print("Metadata DataFrame for: dict_data")
print("-"*100)
dict_metadata_df = pd.DataFrame({
    'Data Type': dict_data.dtypes,
}).reset_index().rename(columns={'index': 'Column Name'})

# Description column
dict_descriptions = {
    "Unnamed: 0": "Index for the diagnosis",
    "diag": "Name of the diagnosis",
    "code": "Diagnostic code (SKS)"
}
dict_metadata_df['Description'] = dict_metadata_df['Column Name'].map(dict_descriptions)
print(dict_metadata_df.to_string())


# Metadata for events 
print("-"*100)
print("Metadata DataFrame for: events")
print("-"*100)
# Create a DataFrame from the metadata
events_metadata_df = pd.DataFrame({
    'Data Type': events.dtypes,
}).reset_index().rename(columns={'index': 'Column Name'})

events_descriptions = {
    "Unnamed: 0": "Index for the diagnosis",
    "id": "Unique identifier for each patient",
    "date": "Date the event occurred",
    "event": "Description of the medical event, e.g Malignancy",
    "type": "Classification of the event type (minor, major, death)"
}
events_metadata_df['Description'] = events_metadata_df['Column Name'].map(events_descriptions)

print(events_metadata_df.to_string())
```

## 2) Translate SKS to ICD-10: In the dict_data.csv dataset, translate the SKS codes into ICD-10 codes. Create a script that checks the SKS code and adds the corresponding ICD-10 code as a new column.
Hint: The ICD-10 codes can be found on this database site: https://icd.who.int/browse10/2019/en.



```python
dict_data
```

```python
icd_codes = []

for code_value in dict_data['code']:
    if code_value == "DE10":
        icd_codes.append('E10')
    elif code_value == "BZFC8A":
        icd_codes.append('E78.5')
    elif code_value == "DE782":
        icd_codes.append('I10')
    else:
            break

dict_data['ICD-10_code'] = icd_codes

dict_data
```
